```python
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/December_projects/cyberbully/1L/cb_multi_labeled_balanced.csv')

df.head()
```

```python
# average word counts

# Calculate the average word count per tweet
df['word_count'] = df['text'].apply(lambda x: len(str(x).split()))
average_word_count = df['word_count'].mean()

print(f"The average word count is: {average_word_count}")
```

```python
# max word count
max_word_count = df['word_count'].max()
print(f"The maximum word count is: {max_word_count}")
```

```python
df['word_count'].describe()
```

```python
# undersampling all class to particular value

from sklearn.utils import resample

# Determine the minimum number of samples per class
min_samples = 15000

# Create an empty list to store the downsampled dataframes
downsampled_dfs = []

# Iterate through each class
for class_label in df['label'].unique():
    # Select the rows belonging to the current class
    class_df = df[df['label'] == class_label]

    # Downsample the current class to the minimum number of samples
    downsampled_class_df = resample(class_df, replace=False, n_samples=min_samples, random_state=42)

    # Append the downsampled dataframe to the list
    downsampled_dfs.append(downsampled_class_df)

# Concatenate all downsampled dataframes into a single dataframe
downsampled_df = pd.concat(downsampled_dfs)

print(downsampled_df['label'].value_counts())
```

```python
df_filtered = downsampled_df.copy()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df_filtered['label'] = le.fit_transform(df_filtered['label'])

# Create the dictionary mapping original labels to encoded labels
label_mapping = dict(zip(le.classes_, le.transform(le.classes_)))

# Print the dictionary
label_mapping
```

```python
# save label_mapping as pickle

import pickle

# Save the label_mapping dictionary as a pickle file
with open('/content/drive/MyDrive/December_projects/cyberbully/label_mapping.pkl', 'wb') as f:
    pickle.dump(label_mapping, f)
```

```python
df_filtered = df_filtered.drop(columns=['word_count'])
df_filtered.head()
```
